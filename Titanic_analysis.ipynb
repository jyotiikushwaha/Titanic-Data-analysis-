```python
#data cleaning 
dataset.isnull().sum()
```

```python
#age has the largest missing value filling with avg value ; 
avg_age=dataset["Age"].mean()
age=dataset["Age"]
age.fillna(avg_age,inplace=True)
dataset["Age"].isnull().sum
```

```python
#handling nan in embarked place ; 

most_visited=dataset["Embarked"].mode()[0]
embarked=dataset["Embarked"]
embarked.fillna(most_visited,inplace=True)
dataset["Embarked"].isnull().value_counts()
```

```python
#cabin nan 
cabin=dataset["Cabin"].mode()[0]
cabin_nan=dataset["Cabin"]
cabin_nan.fillna(cabin,inplace=True)
dataset["Cabin"].isnull().sum()
```

```python
dataset.duplicated().sum()#no duplicates row here 
```

```python
#How many passengers are there?
print(f"Total passenger : {len(dataset)}")
```

```python
#How many males and females are there?
group=dataset.groupby("Sex")
print(len(group))
group.size()
```

```python

#Survival rate by gender?
survived=dataset[dataset["Survived"]==1]
female=survived["Sex"]=="female"
print(female.mean())
male=survived["Sex"]=="male"
print(male.mean())
```

```python
#Survival rate by passenger class?
survival_pclass=dataset.groupby("Pclass")
len(survival_pclass)
survival_pclass["Survived"].mean()
```

```python
#Number of passengers in each class?
passenger=dataset.groupby("Pclass")
passenger.size()
```

```python
#Average age of passengers?
dataset["Age"].mean()
```

```python
#Oldest passenger.
#Youngest passenger.
print(dataset["Age"].max())
print(dataset["Age"].min())
```

```python
#Average fare paid.
#Highest fare.
#Lowest fare.
print(dataset["Fare"].mean())
print(dataset["Fare"].max())
print(dataset["Fare"].min())
```

```python
#Did children survive more often than adults?
dataset["age_distribution"]=np.where(dataset["Age"]<18 ,"Child","Adult")
print(dataset.head(1))#new column add ho gyiii 
print(dataset.groupby("age_distribution")["Survived"].mean()*100)
```

```python
#Which age group survived the most?
sns.histplot(dataset["Age"],color="blue",bins=30,shrink=1.2,edgecolor="black")
plt.title("MOST SURVIVED AGE GROUP")
plt.xlabel("Age group")
plt.ylabel("Survived")
sns.set_style("darkgrid")
plt.show()
```

```python
# between Fare and Survival
print(dataset.groupby("Survived")["Fare"].mean())
sns.boxplot(x=dataset["Survived"],y=dataset["Fare"])
plt.title(str.upper("relationship fare and survival "))
plt.show()
```
